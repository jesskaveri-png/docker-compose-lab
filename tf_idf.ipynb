```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movies = {
'Title':[
'Avengers',
'Captain America',
'Titanic',
'The Notebook',
'John Wick',
'Frozen'
],
'Description':[
'Action movie with superheroes fighting aliens saving world',
'Action superhero fighting war enemies',
'Romantic movie about love on a ship',
'Romantic love story between two people',
'Action assassin revenge fighting enemies',
'Animated movie about two sisters magical powers'
]
}
df = pd.DataFrame(movies)
print(df)
```

```python
tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(df["Description"])
```

```python
print(tfidf.get_feature_names_out())
```

```python
matrix = pd.DataFrame(
tfidf_matrix.toarray(),
columns=tfidf.get_feature_names_out()
)

print(matrix)
```

```python
similarity = cosine_similarity(tfidf_matrix)
```

```python
similarity_df = pd.DataFrame(
similarity,
index=df["Title"],
columns=df["Title"]
)
print(similarity_df)
```

```python
def recommend(movie):
    index = df[df["Title"]==movie].index[0]
    scores = list(enumerate(similarity[index]))
    scores = sorted(scores,key=lambda x:x[1],reverse=True)
    print("Recommended Movies:\n")
    for i in scores[1:4]:
        print(df.iloc[i[0]]["Title"])
```

```python
recommend("Avengers")
```
